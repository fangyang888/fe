## graph 第三阶段 会话

```python

from typing import TypedDict

from langgraph.graph import StateGraph
from langgraph.checkpoint.memory import InMemorySaver


class State (TypedDict):
   count: int

builder = StateGraph(State)

def increment(state: State) -> State:
    return {"count": state["count"] + 1}

builder.add_node("increment", increment)

builder.set_entry_point("increment")
builder.set_finish_point("increment")

graph = builder.compile(checkpointer=InMemorySaver()) # returns {"count": 1}

config = {
    "configurable": {
        "thread_id": "user_001"
    }
}
result = graph.invoke({"count": 0}, config = config) 

print(result)




```

```python
result = graph.invoke({"count": 10}, config = config) 

print(result)
```

```python
result = graph.invoke({"count": 12}, config = config) 

print(result)
```

```python
snapshot = graph.get_state(config)

print("最新状态：", snapshot.values)
```

```python

history = list(graph.get_state_history(config))

print("历史状态：", history)

```

```python

for state in history:
    print("历史状态：", state.values)
```
