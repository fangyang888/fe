```python
from langgraph_py.core.config import openai_settings, anthropic_settings
from langchain.chat_models import init_chat_model
from langchain.tools import tool

model = init_chat_model(
    model_provider="openai",
    model=openai_settings.default_model,
    api_key=openai_settings.api_key,
    base_url=openai_settings.base_url,
)

@tool
def add(a:int,b:int)->int:
    """计算两个整数的和。"""
    return a+b

tools = [add]
model_witch_tools = model.bind_tools(tools)
```

```python
from langchain_core.messages import AIMessage,SystemMessage,HumanMessage

messages = [HumanMessage("2+3等于多少")]
tools_by_name = {tool.name: tool for tool in tools}
for _ in range(5):
 ai_message =  model_witch_tools.invoke(messages)
 messages.append(ai_message)

 if not ai_message.tool_calls:
#   print(ai_message.tool_calls)
  break

 for tool_call in ai_message.tool_calls:
   print(tool_call)
   tool_message = tools_by_name[tool_call["name"]].invoke(tool_call)
   print(tool_message)

```
