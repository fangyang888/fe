```python
import json
from langgraph_py.core.create_sandbox_config import oss_config_settings
metadata_config = {
    "metadata":{
        "fc.sandbox.storage.oss": json.dumps({
            "mountPoints": [
                {
                    "bucketName": oss_config_settings.get("oss_bucket_name"),
                    "mountDir": "/home/user/workspace",
                    "bucketPath": "/e2b-test/workspace",
                    "endpoint": oss_config_settings.get("oss_endpoint"),
                    "readOnly": False,
                },
                {
                    "bucketName": oss_config_settings.get("oss_bucket_name"),
                    "mountDir": "/home/user/output",
                    "bucketPath": "/e2b-test/output",
                    "endpoint": oss_config_settings.get("oss_endpoint"),
                    "readOnly": False,
                }
            ]
        }),
        "fc.sandbox.auth.role": oss_config_settings.get("oss_role_arn"),
    }
}

metadata_config
```

```python
from langgraph_py.core.create_sandbox_config import sandbox_config_settings
from e2b import AsyncSandbox

sandbox = await AsyncSandbox.create(
    **sandbox_config_settings, 
    **metadata_config
)
```

```python
await sandbox.files.write("/home/user/workspace/source.txt", "source")
```

```python
await sandbox.commands.run("node --version")
```

```python
await sandbox.kill()
```
